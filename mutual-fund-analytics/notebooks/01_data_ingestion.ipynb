# Bluestock Mutual Fund Analytics

## 01 - Data Ingestion

This notebook loads the 10 original mutual fund datasets and performs
basic data validation including dataset dimensions, data types,
missing values, duplicate records, and initial observations. 

```python
from pathlib import Path
import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

```python
PROJECT_ROOT = Path.cwd().parent
RAW_DIR = PROJECT_ROOT / "data" / "raw"

print("Project root:", PROJECT_ROOT)
print("Raw data folder:", RAW_DIR)
print("Raw folder exists:", RAW_DIR.exists())
```

```python
csv_files = sorted(
    [
        f for f in RAW_DIR.glob("*.csv")
        if not f.name.endswith("_live_nav.csv")
    ]
)

print("Original datasets found:", len(csv_files))

for file in csv_files:
    print(file.name)
```

```python
datasets = {}

for file in csv_files:
    try:
        df = pd.read_csv(file)
        datasets[file.stem] = df
        print(f"Loaded: {file.name}")
        print(f"Shape: {df.shape}")
        print("-" * 60)
    except Exception as e:
        print(f"ERROR: {file.name}")
        print(e)
```

```python
summary = []

for name, df in datasets.items():
    summary.append({
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "duplicate_rows": int(df.duplicated().sum()),
        "missing_cells": int(df.isna().sum().sum())
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)
    display(df.dtypes.to_frame("data_type"))
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)
    display(df.head())
```

```python
for name, df in datasets.items():
    missing = df.isna().sum()
    missing = missing[missing > 0]

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    if missing.empty:
        print("No missing values.")
    else:
        display(missing.to_frame("missing_count"))
```

```python
duplicate_results = []

for name, df in datasets.items():
    duplicate_results.append({
        "dataset": name,
        "duplicate_rows": int(df.duplicated().sum())
    })

duplicate_df = pd.DataFrame(duplicate_results)

duplicate_df
```

```python
for name, df in datasets.items():
    numeric_cols = df.select_dtypes(include=np.number).columns

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    if len(numeric_cols) > 0:
        display(df[numeric_cols].describe().T)
    else:
        print("No numeric columns.")
```
